```python
# Algoritmo para determinar el mayor de dos cantidades
# Datos de entrada A, B
# Si A es mayor que B, el mayor es A
# Si no, el mayor es B
A = int(input("Ingrese el primer valor (A): "))
B = int(input("Ingrese el segundo valor (B): "))
if A > B:
    M = A
else:
    M = B
print("El mayor es: ", M)
print("Fin del programa")
print("Programa realizado por Hugo NL = 15")
```

```python
# Algoritmo para determinar el mayor de dos cantidades
# Datos de entrada A, B
# Si B es mayor que A, el mayor es B
# Si si, el mayor es A
A = int(input("Ingrese el primer valor (A): "))
B = int(input("Ingrese el segundo valor (B): "))
if B > A:
    M = A
else:
    M = B
print("El mayor es: ", M)
print("Fin del programa")
print("Programa realizado por Hugo NL = 15")
```
